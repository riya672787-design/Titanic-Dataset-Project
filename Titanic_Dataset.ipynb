```python
# Titanic Dataset Load

import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# Outlier Detection using IQR Method

Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Find Age outliers
outliers = df[
    (df["Age"] < lower_bound) |
    (df["Age"] > upper_bound)
]

print("\nNumber of Age Outliers:", len(outliers))
```

```python
print("Number of Age Outliers:", len(outliers))
```

```python
# Outlier Detection using Z-Score Method

mean_age = df["Age"].mean()
std_age = df["Age"].std()

df["Age_ZScore"] = (df["Age"] - mean_age) / std_age

z_outliers = df[df["Age_ZScore"].abs() > 3]

print("Mean Age:", mean_age)
print("Standard Deviation:", std_age)
print("Number of Age Outliers using Z-Score:", len(z_outliers))
```

```python
# Outlier Treatment - Keeping genuine observations

print("Outlier Treatment")
print("------------------")

print("IQR method detected:", len(outliers), "Age outliers")
print("Z-Score method detected: 2 Age outliers")

print("\nDecision: Age outliers are retained.")
print("Reason: Extreme ages can represent genuine passengers.")
print("Removing them may cause loss of useful information.")
```

```python
# Train-Test Split from scratch

import numpy as np

# Target variable
target = "Survived"

# Features and target
X = df.drop(columns=[target])
y = df[target]

# 80% Training, 20% Testing
split_index = int(0.8 * len(df))

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

print("Train-Test Split completed!")
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Data Leakage Prevention
# Calculate preprocessing parameter using TRAINING data only

train_age_mean = X_train["Age"].mean()

print("Training Age Mean:", train_age_mean)

print("\nData Leakage Rule:")
print("Preprocessing parameters are calculated using training data only.")
print("The same training parameter is then applied to test data.")
```

```python
# Sex Encoding again

sex_mapping = {"male": 0, "female": 1}

X_train["Sex_Encoded"] = X_train["Sex"].map(sex_mapping)

print("Sex_Encoded successfully added!")
print(X_train[["Sex", "Sex_Encoded"]].head())
```

```python
# Variance Calculation - Complete

features = ["Pclass", "Age", "SibSp", "Parch", "Fare", "Sex_Encoded"]

for feature in features:
    mean_value = X_train[feature].mean()
    variance = ((X_train[feature] - mean_value) ** 2).mean()

    print(feature, "Variance =", variance)
```

```python
# Variance Threshold - From Scratch

threshold = 0.1

print("Variance Threshold =", threshold)
print("\nFeature Selection Result:")
print("-------------------------")

for feature in features:
    mean_value = X_train[feature].mean()
    variance = ((X_train[feature] - mean_value) ** 2).mean()

    if variance > threshold:
        print(feature, "→ KEEP")
    else:
        print(feature, "→ REMOVE")
```

```python
# Pearson Correlation - From Scratch

features = ["Pclass", "Age", "SibSp", "Parch", "Fare"]

for feature in features:
    x = X_train[feature]
    y = y_train

    x_mean = x.mean()
    y_mean = y.mean()

    numerator = ((x - x_mean) * (y - y_mean)).sum()

    denominator = (
        ((x - x_mean) ** 2).sum() ** 0.5
        * ((y - y_mean) ** 2).sum() ** 0.5
    )

    correlation = numerator / denominator

    print(feature, "Pearson Correlation =", correlation)
```

```python
# Chi-Square Test - From Scratch

feature = "Sex_Encoded"

# Observed frequencies
observed = pd.crosstab(X_train[feature], y_train)

print("Observed Frequency Table:")
print(observed)

# Row and column totals
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)
total = observed.values.sum()

# Calculate Chi-Square
chi_square = 0

for row in observed.index:
    for col in observed.columns:
        expected = (row_totals[row] * col_totals[col]) / total
        observed_value = observed.loc[row, col]

        chi_square += ((observed_value - expected) ** 2) / expected

print("\nChi-Square Value =", chi_square)
```

```python
# ANOVA F-Test - From Scratch

feature = "Age"

group_0 = X_train[y_train == 0][feature]
group_1 = X_train[y_train == 1][feature]

# Overall mean
overall_mean = X_train[feature].mean()

# Group means
mean_0 = group_0.mean()
mean_1 = group_1.mean()

# Number of observations
n0 = len(group_0)
n1 = len(group_1)

# Between-group variation
between = (
    n0 * (mean_0 - overall_mean) ** 2
    + n1 * (mean_1 - overall_mean) ** 2
)

# Within-group variation
within = (
    ((group_0 - mean_0) ** 2).sum()
    + ((group_1 - mean_1) ** 2).sum()
)

# Degrees of freedom
df_between = 1
df_within = n0 + n1 - 2

# Mean squares
MS_between = between / df_between
MS_within = within / df_within

# F-statistic
F_value = MS_between / MS_within

print("ANOVA F-Test for Age vs Survived")
print("---------------------------------")
print("Overall Mean =", overall_mean)
print("Mean Age (Survived = 0) =", mean_0)
print("Mean Age (Survived = 1) =", mean_1)
print("Between-group variation =", between)
print("Within-group variation =", within)
print("F-statistic =", F_value)
```

```python
# Mutual Information (MI) - From Scratch

feature = "Sex_Encoded"

# Total number of observations
total = len(X_train)

# Entropy of Y (Survived)
y_counts = y_train.value_counts()

H_Y = 0

for count in y_counts:
    p = count / total
    H_Y -= p * np.log2(p)

# Conditional Entropy H(Y|X)
H_Y_given_X = 0

x_values = X_train[feature].unique()

for x_value in x_values:

    subset_y = y_train[X_train[feature] == x_value]
    n_x = len(subset_y)

    y_counts_x = subset_y.value_counts()

    entropy_x = 0

    for count in y_counts_x:
        p = count / n_x
        entropy_x -= p * np.log2(p)

    H_Y_given_X += (n_x / total) * entropy_x

# Mutual Information
MI = H_Y - H_Y_given_X

print("Mutual Information - From Scratch")
print("-----------------------------------")
print("Feature:", feature)
print("Entropy H(Y) =", H_Y)
print("Conditional Entropy H(Y|X) =", H_Y_given_X)
print("Mutual Information =", MI)
print("Unit: bits")
```

```python
print("Mutual Information Interpretation")
print("----------------------------------")

print("Feature:", feature)
print("MI Value:", MI)

if MI > 0:
    print("Conclusion: Sex_Encoded contains information about Survived.")
else:
    print("Conclusion: No measurable information was found.")
```

```python
# Feature Selection - Method Comparison

print("FEATURE SELECTION METHOD COMPARISON")
print("------------------------------------")

print("1. Variance Threshold → Low-variance features are removed")
print("2. Pearson Correlation → Measures linear relationship with target")
print("3. Chi-Square → Measures relationship between categorical feature and target")
print("4. ANOVA F-Test → Measures difference between numerical feature groups")
print("5. Mutual Information → Measures information/dependence between feature and target")

print("\nImportant Result:")
print("Sex_Encoded MI =", MI, "bits")
print("Sex_Encoded shows useful information about Survived.")
```

```python
# ============================================
# Cost Function J(theta0, theta1)
# ============================================

import numpy as np
import matplotlib.pyplot as plt

# Titanic data
x = df["Fare"].fillna(df["Fare"].median()).values
y = df["Survived"].values

# Normalize Fare for better visualization
x_norm = (x - x.mean()) / x.std()

# Cost Function
def compute_cost(theta0, theta1, x, y):
    m = len(y)

    h = theta0 + theta1 * x

    cost = (1 / (2 * m)) * np.sum((h - y) ** 2)

    return cost


# ============================================
# Create theta values
# ============================================

theta0_values = np.linspace(-2, 2, 100)
theta1_values = np.linspace(-2, 2, 100)

J = np.zeros((len(theta0_values), len(theta1_values)))


# Calculate Cost
for i, theta0 in enumerate(theta0_values):
    for j, theta1 in enumerate(theta1_values):
        J[i, j] = compute_cost(theta0, theta1, x_norm, y)


# ============================================
# Contour Plot
# ============================================

plt.figure(figsize=(9, 6))

plt.contour(
    theta0_values,
    theta1_values,
    J.T,
    levels=20
)

plt.xlabel("θ₀ (Theta 0)")
plt.ylabel("θ₁ (Theta 1)")
plt.title("Cost Function J(θ₀, θ₁) - Titanic Dataset")

plt.grid(True)

plt.colorbar(label="Cost J")

plt.show()
```

```python
# ==========================================
# Gradient Descent Visualization - Titanic
# ==========================================

import numpy as np
import matplotlib.pyplot as plt

# Normalized Fare aur Survival
x = df["Fare"].fillna(df["Fare"].median()).values
y = df["Survived"].values

# Normalize Fare
x = (x - x.mean()) / x.std()

# Initial parameters
theta0 = 2.0
theta1 = 2.0

# Learning rate
alpha = 0.05

# Number of iterations
iterations = 100

# Store theta values
theta0_history = []
theta1_history = []
cost_history = []

# Gradient Descent
m = len(y)

for i in range(iterations):

    # Prediction
    h = theta0 + theta1 * x

    # Cost
    J = (1 / (2 * m)) * np.sum((h - y) ** 2)

    # Gradients
    dtheta0 = (1 / m) * np.sum(h - y)
    dtheta1 = (1 / m) * np.sum((h - y) * x)

    # Store values
    theta0_history.append(theta0)
    theta1_history.append(theta1)
    cost_history.append(J)

    # Update parameters
    theta0 = theta0 - alpha * dtheta0
    theta1 = theta1 - alpha * dtheta1

print("Gradient Descent completed successfully!")
print("Final Theta 0:", theta0)
print("Final Theta 1:", theta1)
print("Final Cost:", cost_history[-1])
```

```python
# Cost Function vs Iterations

plt.figure(figsize=(8, 5))

plt.plot(cost_history, linewidth=2)

plt.xlabel("Iterations")
plt.ylabel("Cost J(θ)")
plt.title("Cost Function vs Iterations - Gradient Descent")

plt.grid(True)
plt.show()
```
